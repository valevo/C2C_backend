```python
import pandas as pd

import langid
# langid.classify("Bonjour le monde")  # ('fr', -54.41)
def langid_classify(t):
    return langid.classify(t)[0]
```

```python
cols = ["timestamp", "text", "permission_processing", "permission_sharing"]
short_nl = pd.read_csv("./responses_short_NL_20260821.csv")
short_nl.columns = cols
short_en = pd.read_csv("./responses_short_EN_20260821.csv")
short_en.columns = cols

short_column_ids = [0, 1, -2, -1]

nl = pd.read_csv("./responses_NL_20260821.csv")
nl = nl.iloc[:, short_column_ids]
nl.columns = cols
en = pd.read_csv("./responses_EN_20260821.csv")
en = en.iloc[:, short_column_ids]
en.columns = cols


combined = pd.concat([short_nl, nl, short_en, en], ignore_index=True)
# combined.timestamp = pd.to_datetime(combined.timestamp, format="%d/%m/%Y %H:%M:%S")
langs = combined.text.apply(langid_classify)
combined["language_code"] = langs
combined = combined.drop_duplicates("text")
```

```python
combined.to_csv("conversation_starters_20260821.csv")
```

```python
langs = short_nl.text.apply(langid_classify)
short_nl["language_code"] = langs

langs = short_en.text.apply(langid_classify)
short_en["language_code"] = langs

short = pd.concat([short_nl, short_en])
short.timestamp = pd.to_datetime(short.timestamp, format="%d/%m/%Y %H:%M:%S")
short
```
